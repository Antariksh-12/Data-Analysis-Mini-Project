# Data Visualization Mini Project : 120 Years of Olympic History

# Introduction 

#### The dataset used for this analysis is the "Olympic Games 1896-2016" dataset (`athlete_events.csv`). It contains historical data on athletes, including their physical attributes, nationality, sport, and medal achievements.
#### The objectivbe of the analysis is to clean ,aggregate and visualize historical Olympic data to discover trends in Country participation, gender distribution, physical attributes of atheletes correlations between age, height and weight. 



## Dataset Description 

### Source - Open source Olympics historical dataset.
### Important Variables Used - `Name`, `Sex`, `Age`, `Height`, `Weight`, `NOC` (Country Code), `Year`, `Season`, `Sport`, and `Medal`.

```python
# Importing necessary libraries for data manipulation and visualization
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set a clean visual style for all our plots and a cleaner output 
sns.set_theme(style="whitegrid")
import warnings
warnings.filterwarnings('ignore')
```

## Data Processing 

## In this step we will :
1. Load the dataset.
2. Handle missing values (e.g., replacing NaN in the 'Medal' column with 'No Medal').
3. Filter the data to focus exclusively on the **Summer Olympics** to make our historical trends consistent.

```python
df = pd.read_csv('athlete_events.csv')

# Handling missing values
# If an athlete didn't win a medal, the value is NaN. We fill it with 'No Medal'
df['Medal'] = df['Medal'].fillna('No Medal')

# Data Cleaning
# Drop exact duplicate rows if any exist
df = df.drop_duplicates()

# Filtering
# We filter the dataset to only analyze the Summer Olympics
summer_df = df[df['Season'] == 'Summer']

# Display the first 10 rows to confirm preprocessing
summer_df.head(10)
```

## Visualization and Analysis 

### Plot 1: Top 10 Participating Countries (Bar Chart)
We aggregate the total number of athletes sent by each country over the years.


```python
plt.figure(figsize=(10,5))
# Grouping by Country Code (NOC) and taking the top 10
top_countries = summer_df['NOC'].value_counts().head(10)

sns.barplot(x=top_countries.index, y=top_countries.values, palette="viridis")
plt.title('Top 10 Countries by Total Athlete Participation')
plt.xlabel('Country (NOC Code)')
plt.ylabel('Number of Athletes')
plt.show()
```

## Insight:
#### The United States (USA), France (FRA), and Great Britain (GBR) have historically sent the most athletes to the Summer Olympics.

## Plot 2: Gender Distribution (Bar Chart)
 Let's look at the overall comparison of male vs. female participants.

```python
plt.figure(figsize=(7,5))
sns.countplot(x='Sex', data=summer_df, palette='Set2')
plt.title('Total Male vs Female Participation (1896-2016)')
plt.xlabel('Gender (M = Male, F = Female)')
plt.ylabel('Total Count')
plt.show()
```

## Insight:
#### Historically, there have been significantly more male athletes than female athletes overall across the 120 years.

## Plot 3 and 4 : Historical Trends (Line Charts)


How has the scale of the Olympics grown over time in terms of athletes and participating nations?

```python
# Grouping data by Year to find unique athletes and unique nations
athletes_over_time = summer_df.groupby('Year')['ID'].nunique()
nations_over_time = summer_df.groupby('Year')['NOC'].nunique()

# Creating a figure with 2 subplots side by side
fig, ax = plt.subplots(1, 2, figsize=(16, 5))

# Plot 3: Athletes over time
ax[0].plot(athletes_over_time.index, athletes_over_time.values, marker='o', color='blue')
ax[0].set_title('Growth of Athlete Participation Over Time')
ax[0].set_xlabel('Year')
ax[0].set_ylabel('Number of Unique Athletes')

# Plot 4: Nations over time
ax[1].plot(nations_over_time.index, nations_over_time.values, marker='s', color='green')
ax[1].set_title('Growth of Participating Nations Over Time')
ax[1].set_xlabel('Year')
ax[1].set_ylabel('Number of Nations')

plt.show()
```

## Insight:
#### Both the number of athletes and participating nations show an upward trend, proving the Olympics have become a massive global event. The sharp drops correspond to the World Wars.



## Plot 5 and 6 : Physical Attributes (Histograms)


 Observing the distribution of the athletes' ages and heights.

```python
fig, ax = plt.subplots(1, 2, figsize=(16, 5))

# Plot 5: Age Distribution
sns.histplot(summer_df['Age'].dropna(), bins=25, color='purple', kde=False, ax=ax[0])
ax[0].set_title('Distribution of Athlete Ages')
ax[0].set_xlabel('Age')

# Plot 6: Height Distribution
sns.histplot(summer_df['Height'].dropna(), bins=25, color='orange', kde=True, ax=ax[1])
ax[1].set_title('Distribution of Athlete Heights')
ax[1].set_xlabel('Height (cm)')

plt.show()
```

## Insight :
#### Most Olympic athletes are in their early-to-mid 20s. The height distribution forms a nearly perfect normal distribution (bell curve) centered around 175 cm.


## Plot 7 & 8: Height vs. Weight (Scatter Plots)

How do height and weight correlate, and how does it differ in specific sports like Basketball?

```python
fig, ax = plt.subplots(1, 2, figsize=(16, 6))

# Plot 7: General Height vs Weight colored by Sex
sns.scatterplot(x='Height', y='Weight', data=summer_df, hue='Sex', alpha=0.5, ax=ax[0])
ax[0].set_title('Height vs Weight (All Athletes)')

# Plot 8: Height vs Weight for Basketball only
basketball_df = summer_df[summer_df['Sport'] == 'Basketball']
sns.scatterplot(x='Height', y='Weight', data=basketball_df, hue='Sex', palette='husl', ax=ax[1])
ax[1].set_title('Height vs Weight (Basketball Players Only)')

plt.show()
```

## Insight :
#### There is a clear positive correlation between height and weight. Males generally cluster higher on both axes. Basketball players are distinctly taller and heavier than the average athlete.

## Plot 9 : Age across Medal Winners (Box Plot)

Does age vary depending on the type of medal won?


```python
plt.figure(figsize=(8,5))
# Filter out non-medalists for this plot
medal_winners = summer_df[summer_df['Medal'] != 'No Medal']

sns.boxplot(x='Medal', y='Age', data=medal_winners, order=['Gold', 'Silver', 'Bronze'], palette='pastel')
plt.title('Age Distribution Across Medal Winners')
plt.show()
```

## Insight :
#### The median age for Gold, Silver, and Bronze medalists is virtually identical (around 24-25 years old). However, there are older outliers (athletes in their 50s and 60s) winning medals, likely in sports like Equestrian or Shooting.


## Plot 10 : Attribute Correlation (Heatmap)

A mathematical look at how Age, Height, and Weight are related.

```python
plt.figure(figsize=(7,5))
# Select only numeric columns for correlation
correlation_data = summer_df[['Age', 'Height', 'Weight']].corr()

sns.heatmap(correlation_data, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap: Age, Height, and Weight')
plt.show()
```

## Insight :
#### Height and Weight are highly positively correlated (0.80), which makes physical sense. Age has a very weak correlation with height and weight, meaning physical size does not strictly increase with age for adult Olympians."
